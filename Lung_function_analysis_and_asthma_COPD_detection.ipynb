```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# ✅ STEP 1: Install Gradio
!pip install -q gradio

# ✅ STEP 2: Import Required Libraries
import gradio as gr

# ✅ STEP 3: Define Rule-Based Diagnostic Function
def diagnose_lung(FEV1, FVC):
    try:
        ratio = FEV1 / FVC
    except ZeroDivisionError:
        return "Invalid input: FVC cannot be zero."

    diagnosis = ""
    explanation = f"FEV1/FVC ratio = {ratio:.2f}\n"

    if ratio >= 0.70 and FEV1 >= 80:
        diagnosis = "Normal Lung Function"
    elif ratio < 0.70 and FEV1 < 80:
        diagnosis = "Chronic Obstructive Pulmonary Disease (COPD)"
    elif ratio < 0.70 and FEV1 >= 80:
        diagnosis = "Asthma (Possible - requires bronchodilator test to confirm)"
    else:
        diagnosis = "Uncertain - Further tests needed"

    return f"{diagnosis}\n\n{explanation}FEV1 = {FEV1}%, FVC = {FVC}"

# ✅ STEP 4: Create Gradio Interface
iface = gr.Interface(
    fn=diagnose_lung,
    inputs=[
        gr.Number(label="FEV1 (% Predicted)"),
        gr.Number(label="FVC")
    ],
    outputs="text",
    title="Lung Function Analyzer",
    description="Enter spirometry results (FEV1 and FVC) to detect Normal lung function, Asthma, or COPD based on medical thresholds."
)

# ✅ STEP 5: Launch the Web App
iface.launch()
```
